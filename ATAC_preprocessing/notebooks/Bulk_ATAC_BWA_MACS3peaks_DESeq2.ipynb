```r
runDESeq2 <- function(atacDDS, grouping="Group", groups, format = "GRanges", sub = FALSE){
    if(sub){
        atacDDS <- atacDDS[,which(atacDDS$Group %in% groups)]
        atacDDS$Group <- droplevels(atacDDS$Group)
        atacDDS <- estimateSizeFactors(atacDDS)
        atacDDS <- DESeq(atacDDS)
    }
    res <- results(atacDDS,c(grouping,groups[1],groups[2]),format=format)
    return(res)
}
```

### Function to filter peaks by log2FoldChange and padj.

```r
filter_fdr_lfc <- function(a, abs = TRUE, lfc_cutoff = 0.5, fdr = TRUE,
                           fdr_cutoff = 0.20, OR = FALSE,
                           returnFilteredPeaks = TRUE){
    
    require(GenomicAlignments)
    
    x <- a[!(is.na(a$padj))]
    y <- x # y: used to filter by fdr
    z <- x # z: used to filter by logFC
    
    if(fdr){
        y <- y[y$padj > fdr_cutoff]
    if(abs){
        z <- z[abs(z$log2FoldChange) < lfc_cutoff]
    } else {
        z <- z[z$log2FoldChange < lfc_cutoff]
    }
    }

    if(OR){
        grl <- GRangesList(y,z)
        gr <- reduce(unlist(grl))
    } else{
        gr <- y[y %over% z]
    }

    if(returnFilteredPeaks){
        if(abs){
            if(OR){
                print(paste0("Returning filtered peaks: FDR > ", fdr_cutoff, " or |logFC| < ", lfc_cutoff))
            } else{
                print(paste0("Returning filtered peaks: FDR > ", fdr_cutoff, " and |logFC| < ", lfc_cutoff))
            }
        } else{
            if(OR){
                print(paste0("Returning filtered peaks: FDR > ", fdr_cutoff, " or logFC < ", lfc_cutoff))
            } else{
                print(paste0("Returning filtered peaks: FDR > ", fdr_cutoff, " and logFC < ", lfc_cutoff))
            }
        }
                  
        a <- gr
                  
    } else{
        if(abs){
            if(OR){
                print(paste0("Returning peaks to filter: FDR = NA or < ", fdr_cutoff, " or |logFC| > ", lfc_cutoff))
            } else{
                print(paste0("Returning peaks to filter: FDR = NA or < ", fdr_cutoff, " and |logFC| > ", lfc_cutoff))
            }
        } else{
            if(OR){
                print(paste0("Returning peaks to filter: FDR = NA or < ", fdr_cutoff, " or logFC > ", lfc_cutoff))
            } else{
                print(paste0("Returning peaks to filter: FDR = NA or < ", fdr_cutoff, " and logFC > ", lfc_cutoff))
            }
        }
        
        a <- a[!a %over% gr] # note: includes padj = NA peaks
    }
    return(a)
}
```

#### Get D30-resolved peaks.

```r
# Take an equal number of peaks from each side of the zero reference for "zero-reference regions".
# If one side has an extra peak, remove this peak so that each side (i.e. left/right
# of the zero reference) has an equal number of peaks.
num_unchanged <- length(unchanged)
num_total <- nrow(d30_DF_res_stat_ordered)

if(num_total %% 2 == 0){
  if(num_unchanged %% 2 == 0){
    lower_limit_zeroref <- zero_reference - (num_unchanged/2) - 1
    upper_limit_zeroref <- zero_reference + (num_unchanged/2)
  } else{
    num_unchanged_new <- num_unchanged - 1
    lower_limit_zeroref <- zero_reference - (num_unchanged_new/2) - 1
    upper_limit_zeroref <- zero_reference + (num_unchanged_new/2)
  }
} else{
  if(num_unchanged %% 2 == 1){
    lower_limit_zeroref <- zero_reference - as.integer(num_unchanged/2) - 1
    upper_limit_zeroref <- zero_reference + as.integer(num_unchanged/2) + 1
  } else{
    num_unchanged_new <- num_unchanged - 1
    lower_limit_zeroref <- zero_reference - as.integer(num_unchanged_new/2) - 1
    upper_limit_zeroref <- zero_reference + as.integer(num_unchanged_new/2) + 1
  }
}

if(zero_reference < as.integer(num_unchanged/2) - 1){
    lower_limit_zeroref <- 0
    upper_limit_zeroref <- num_unchanged + 1
}
```

### Calculate mean and standard error for each peak set, condition, and timepoint.

```r
colnames(normalized_counts)
```

```r
calculate_stats <- function(dds, peak_set, counts, sample_info) {
  # Get the indices of peaks in the count matrix that overlap with the peak set
  overlaps <- findOverlaps(peak_set, rowRanges(dds))
  peak_indices <- unique(subjectHits(overlaps))
  
  counts_subset <- counts[peak_indices, ]
  
  # Convert sample_info to a regular data frame if it's not already
  if (is(sample_info, "DFrame")) {
    sample_info <- as.data.frame(sample_info)
  }
  
  # Ensure sample names are a column in sample_info
  if (!"Sample" %in% colnames(sample_info)) {
    sample_info$Sample <- rownames(sample_info)
  }
  
  stats <- counts_subset %>%
    as.data.frame() %>%
    tibble::rownames_to_column("Peak") %>%
    pivot_longer(-Peak, names_to = "Sample", values_to = "Count") %>%
    left_join(sample_info, by = "Sample") %>%
    group_by(Condition, Timepoint) %>%
    summarize(
      mean_signal = mean(log2(Count + 1)),
      se_signal = sd(log2(Count + 1)) / sqrt(n()),
      .groups = "drop"
    )
  
  # Return both summary stats and individual replicate data
  list(
    summary = stats,
    replicates = counts_subset %>%
      as.data.frame() %>%
      tibble::rownames_to_column("Peak") %>%
      pivot_longer(-Peak, names_to = "Sample", values_to = "Count") %>%
      left_join(sample_info, by = "Sample") %>%
      group_by(Condition, Timepoint, Sample) %>%
      summarize(
        mean_signal = mean(log2(Count + 1)),
        .groups = "drop"
      )
  )
}
```

```r
# Get stats over each peak set
d30_memory_stats <- calculate_stats(dds=atacDDS_all, peak_set=d30_memory, counts=normalized_counts, sample_info=colData(atacDDS_all))
resolved_stats <- calculate_stats(dds=atacDDS_all, peak_set=d30_resolved, counts=normalized_counts, sample_info=colData(atacDDS_all))
unchanged_stats <- calculate_stats(dds=atacDDS_all, peak_set=unchanged, counts = normalized_counts, sample_info=colData(atacDDS_all))

m12_resolved_memory_stats <- calculate_stats(dds=atacDDS_all, peak_set=m12_resolved_memory, counts = normalized_counts, sample_info=colData(atacDDS_all))
m12_memory_stats <- calculate_stats(dds=atacDDS_all, peak_set=m12_memory, counts = normalized_counts, sample_info=colData(atacDDS_all))
```

```r
# Combine statistics for all peak sets
all_stats_summary_01 <- bind_rows(
    d30_memory_stats$summary %>% mutate(peak_set = "Memory"),
    resolved_stats$summary %>% mutate(peak_set = "Resolved"),
    unchanged_stats$summary %>% mutate(peak_set = "Unchanged")
)

all_stats_summary_02 <- bind_rows(
    m12_resolved_memory_stats$summary %>% mutate(peak_set = "Short-term Memory"),
    m12_memory_stats$summary %>% mutate(peak_set = "Long-term Memory"),
    # resolved_stats$summary %>% mutate(peak_set = "Resolved"),
    # unchanged_stats$summary %>% mutate(peak_set = "Unchanged")
)
```

```r
# Combine individual replicate data for all peak sets
all_stats_replicates_01 <- bind_rows(
    d30_memory_stats$replicates %>% mutate(peak_set = "Memory"),
    resolved_stats$replicates %>% mutate(peak_set = "Resolved"),
    unchanged_stats$replicates %>% mutate(peak_set = "Unchanged")
)

all_stats_replicates_02 <- bind_rows(
    m12_resolved_memory_stats$replicates %>% mutate(peak_set = "Short-term Memory"),
    m12_memory_stats$replicates %>% mutate(peak_set = "Long-term Memory"),
    # resolved_stats$replicates %>% mutate(peak_set = "Resolved"),
    # unchanged_stats$replicates %>% mutate(peak_set = "Unchanged")
)
```

```r
timepoint_order <- c("D6","D30","D180","M12","M18","M24")
```

```r
# Create the plot
ggplot() +
  geom_ribbon(data = all_stats_summary_01, 
              aes(x = factor(Timepoint, levels = timepoint_order), 
                  ymin = mean_signal - se_signal, 
                  ymax = mean_signal + se_signal, 
                  fill = Condition, 
                  group = interaction(peak_set, Condition)), 
              alpha = 0.3) +
  geom_jitter(data = all_stats_replicates_01, 
              aes(x = factor(Timepoint, levels = timepoint_order), 
                  y = mean_signal, 
                  color = Condition),
              size = 2.5, alpha = 0.8,
              width = 0.1, height = 0) +
  facet_wrap(~ peak_set) +
  labs(x = "Timepoint", y = "log2(DESeq2 ATAC-seq signal)") +
  ylim(3,8) +
  theme_minimal() +
  scale_color_manual(values = c("IMQ" = "red","Ctrl" = "black")) +
  scale_fill_manual(values = c("IMQ" = "red","Ctrl" = "black")) +
  theme(legend.position = "bottom")

# Save the plot
if(!dir.exists("/rugpfs/fs0/fuchs_lab/scratch/ssajjath/Aging_ATAC/final_outs/signal_plots/")){
    dir.create("/rugpfs/fs0/fuchs_lab/scratch/ssajjath/Aging_ATAC/final_outs/signal_plots/")
}
ggsave("/rugpfs/fs0/fuchs_lab/scratch/ssajjath/Aging_ATAC/final_outs/signal_plots/atac_seq_signal_plot_01.pdf", width = 12, height = 8)
```
